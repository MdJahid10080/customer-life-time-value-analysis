```python
snapshot_date = df['InvoiceDate'].max() + pd.Timedelta(days=1)
```

```python
recency = (
    df.groupby('CustomerID')['InvoiceDate']
      .max()
)
```

```python
recency = (snapshot_date - recency).dt.days
```

```python
recency.head()
```

```python
rfm = pd.DataFrame({
    'Recency': recency,
    'Frequency':purchase_frequency,
    'Monetary': customer_revenue
})
```

```python
rfm.head()
```

```python
rfm['R_Score'] = pd.qcut(
    rfm['Recency'],
    q=5,
    labels=[5,4,3,2,1]
)
```

```python
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'],
    q=5,
    labels=False,
    duplicates='drop'
) + 1
```

```python
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'],
    q=5,
    labels=[1,2,3,4,5]
)
```

```python
rfm.head()
```

```python
print(rfm.dtypes)
```

```python
rfm['R_Score'] = rfm['R_Score'].astype(int)
rfm['M_Score'] = rfm['M_Score'].astype(int)
```

```python
print(rfm.dtypes)
```

```python
rfm['RFM_Score'] = (
    rfm['R_Score'].astype(str) +
    rfm['F_Score'].astype(str) +
    rfm['M_Score'].astype(str)
)
```

```python
rfm.head()
```

```python
def customer_segment(row):

    if row['R_Score'] >= 4 and row['F_Score'] >= 4:
        return 'Champions'

    elif row['R_Score'] >= 3 and row['F_Score'] >= 3:
        return 'Loyal Customers'

    elif row['R_Score'] >= 4 and row['F_Score'] <= 2:
        return 'Potential Loyalists'

    elif row['R_Score'] <= 2 and row['F_Score'] >= 3:
        return 'At Risk'

    else:
        return 'Lost Customers'
```

```python
rfm['Segment'] = rfm.apply(customer_segment, axis=1)
```

```python
rfm['Average_Order_Value'] = rfm['Monetary'] / rfm['Frequency']
```

```python
#MACHINE LEARNING
# Features (Input)
X = rfm[['Recency', 'Frequency', 'Monetary']]

# Target (Output)
y = rfm['Segment']
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder object
label_encoder = LabelEncoder()

# Encode target variable
y_encoded = label_encoder.fit_transform(y)
```

```python
print(label_encoder.classes_)
```

```python
print(y_encoded[:10])
```

```python
mapping = dict(zip(label_encoder.classes_,
                   label_encoder.transform(label_encoder.classes_)))

print(mapping)
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create the model
rf_model = RandomForestClassifier(
    n_estimators=100,# it include 100 decision tree
    random_state=42
)

# Train the model
rf_model.fit(X_train, y_train)
```

```python
print(rf_model)
```

```python
# Predict customer segments
y_pred = rf_model.predict(X_test)
```

```python
print("Actual Labels:")
print(y_test[:10])

print("\nPredicted Labels:")
print(y_pred[:10])
```

```python
# converting them back to customer segment names.
predicted_segments = label_encoder.inverse_transform(y_pred)

print(predicted_segments[:10])
```

```python
#ACCURACY
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy*100,"%")
```

```python
#PRECISION
from sklearn.metrics import precision_score

precision = precision_score(
    y_test,
    y_pred,
    average='weighted'
)

print("Precision:", precision*100,"%")
```

```python
# RECALL
from sklearn.metrics import recall_score

recall = recall_score(
    y_test,
    y_pred,
    average='weighted'
)

print("Recall:", recall*100,"%")
```

```python
from sklearn.metrics import f1_score

f1 = f1_score(
    y_test,
    y_pred,
    average='weighted'
)

print("F1 Score:", f1)
```

```python
#confussion matrics
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

disp.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
))
```

```python
# Get feature importance
importance = pd.DataFrame({
    'Feature': X.columns,
    'Importance': rf_model.feature_importances_
})

# Sort from highest to lowest
importance = importance.sort_values(
    by='Importance',
    ascending=False
)

print(importance)
```

```python
plt.figure(figsize=(7,5))

plt.bar(
    importance['Feature'],
    importance['Importance']
)

plt.title("Feature Importance")
plt.xlabel("Features")
plt.ylabel("Importance Score")

plt.show()
```
